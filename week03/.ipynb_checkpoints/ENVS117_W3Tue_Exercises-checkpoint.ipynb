## E1. Reservoir storage check (Built-ins, water resources)

````python
# Five California reservoirs. Units: thousand acre-feet (TAF).
# Capacities are published values; storage numbers are synthetic for practice.
reservoirs = ["Shasta", "Oroville", "Folsom", "San Luis", "Don Pedro"]
capacity_taf = [4552, 3537, 977, 2041, 2030]
storage_raw = ["2810", "1985.5", "402", "690", "1460"]   # read from a report, so it is text
````

**1.** Convert `storage_raw` to a list of floats called `storage_taf`.

````python
# your code here
i = 0
storage_taf = list()
while i < len(storage_raw):
    storage_taf.append(float(storage_raw[i]))
    i = i + 1

````

**2.** With `zip`, compute percent full for each reservoir, rounded to 1 decimal. Store the values in a list `pct` and print a neat table.

````python
# your code here
pct = list()
for storage, capacity in zip(storage_taf, capacity_taf):
    pct.append(round(storage/capacity * 100, 1))
    
for name, percent in zip(reservoirs, pct):
    print(f"{name:<12} {percent:5.1f}")
````

**3.** Name the fullest and the emptiest reservoir using `max`, `min` and `.index()`.

````python
# your code here

print("Fullest reservoir:\n", reservoirs[pct.index(max(pct))], "\nPercent:",  max(pct))
print("Emptiest reservoir:\n", reservoirs[pct.index(min(pct))], "\nPercent:",  min(pct))


````

**4.** Use `sum` to find the percent full of all five reservoirs together.

````python
# your code here
capacity_sum = sum(capacity_taf)
storage_sum = sum(storage_taf)
pct_combined = storage_sum / capacity_sum * 100
print(pct_combined)
````

**5.** Answer with `any` and `all`: is any reservoir below 40%? Are all above 30%?

````python
# your code here
below_40pct = any(i < 40 for i in pct)
above_30pct = all(i > 30 for i in pct)

print(below_40pct)
print(above_30pct)
````

**6.** Loop with `enumerate(start=1)` and label each reservoir `"low"` if under 40%, else `"ok"`, using a one-line if / else.

````python
# your code here
status = None
for i, (name, percent) in enumerate(zip(reservoirs, pct), start=1):
    status = "low" if percent < 40 else "ok"
    print(f"{i}. {name}: {status}")
````
