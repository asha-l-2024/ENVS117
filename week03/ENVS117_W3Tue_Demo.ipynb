## Part 1a. Built-in functions: numbers and types
Built-ins need no import. Full list: https://docs.python.org/3/library/functions.html

````python
print(abs(-8))           # 8
print(float("8.0"))      # 8.0
print(int("8"))          # 8
print(pow(4, 3))         # 64
print(4 ** 3)            # same as pow(4, 3)
print(round(2.36))       # 2  (no second argument: round to an integer)
print(round(2.36, 1))    # 2.4
print(str(10), type(str(10)))   # "10" is now text
````

Why conversion matters for us: anything read from a CSV or text file arrives as a **string**.

````python
pm_from_file = "36.2"          # text, as read from a file
print(type(pm_from_file))

pm = float(pm_from_file)       # convert before doing math
print(pm + 1.0, type(pm))
````

## Part 1b. Built-ins for lists
Synthetic PM2.5 values from last week's Smoke Event stations.

````python
pm25 = [12.4, 38.1, 9.0, 61.7]
names = ["SJ", "Gilroy", "SCU", "Morgan Hill"]

print(len(pm25))                       # 4
print(min(pm25), max(pm25))            # 9.0 61.7
print(sum(pm25))                       # 121.2
print(round(sum(pm25) / len(pm25), 1)) # mean: 30.3
````

````python
# sorted() returns a NEW list; the original is unchanged
print(sorted(pm25))
print(sorted(pm25, reverse=True))
print(pm25)
````

````python
# zip pairs items from two lists
for name, pm in zip(names, pm25):
    print(f"{name:<12} {pm:5.1f}")
````

````python
# enumerate gives the position and the item
for i, name in enumerate(names, start=1):
    print(i, name)
````

````python
# any / all with a condition inside
print(any(pm > 35.4 for pm in pm25))   # is ANY station above the sensitive-groups threshold?
print(all(pm < 100 for pm in pm25))    # are ALL stations below 100?
````

````python
# which station is highest? max + index
worst = names[pm25.index(max(pm25))]
print("Highest PM2.5:", worst)
````

````python
# what kind of object is it?
print(type(pm25), type(pm25[0]), type(names[0]))
print(isinstance(pm25[0], float))
````

## Part 1c. Three surprises (*Predict first*)

````python
# Surprise 1: int() cannot read decimal text
try:
    int("8.0")
except ValueError as e:
    print("ValueError:", e)

print(int(float("8.0")))     # the fix: go through float first
````

````python
# Surprise 2: exact halves round to the EVEN neighbor
print(round(2.5))   # 2
print(round(3.5))   # 4
print(round(0.5), round(1.5))   # 0 2
````

````python
# Surprise 3: floats carry tiny binary errors
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)

import math
print(math.isclose(0.1 + 0.2, 0.3))   # compare floats with a tolerance
````

## Part 1d. Comparison operators
Every comparison answers `True` or `False`.

````python
print(4 == 9)    # False
print(4 != 9)    # True
print(4 > 9)     # False
print(4 < 9)     # True
print(3 >= 3)    # True
print(3 <= 2)    # False
````

````python
# a comparison is a value: you can store it
pm = 36.2
is_smoky = pm > 35.4
print(is_smoky, type(is_smoky))

# = assigns, == compares
region = "West"
print(region == "West")
````

````python
# text compares exactly, including capital letters
name = "Shasta"
print(name == "shasta")             # False
print(name.lower() == "shasta")     # True

messy = "  SHASTA "
print(messy.strip().lower() == "shasta")   # clean first, then compare
````

````python
# count how many stations are above a threshold: True counts as 1
print(sum(pm > 35.4 for pm in pm25))
````

## Part 1e. Combining conditions

````python
lat, lon = 37.35, -121.94      # Santa Clara University

in_range = 32.5 <= lat <= 42.0                 # chained range check
print("In California latitude range:", in_range)

valid = -90 <= lat <= 90 and -180 <= lon <= 180  # and: both must be True
print("Valid coordinate:", valid)
````

````python
# membership with in
county = "Santa Clara"
tracked = county in ["Santa Clara", "Alameda"]
print(tracked)

print("lake" in ["Lake Mead", "Shasta"])   # False: in checks WHOLE items
print("Lake" in "Lake Mead")              # True: in on a string checks text inside it
````

````python
# one-line if / else (conditional expression)
pct = 33.8
status = "low" if pct < 40 else "ok"
print(status)
````

````python
# empty things count as False
readings = []
if not readings:
    print("no data")

print(bool([]), bool(""), bool(0), bool(None))   # all False
print(bool([0]), bool("0"), bool(-1))            # all True
````

## Quick check (*Predict first*)
A CSV gives you `pm = "36.2"`. Is `pm > 35.4` True?

````python
pm = "36.2"
try:
    print(pm > 35.4)
except TypeError as e:
    print("TypeError:", e)

print(float(pm) > 35.4)    # convert first: True
````
