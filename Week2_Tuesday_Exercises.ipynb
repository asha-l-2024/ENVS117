## T5 (stretch). Reservoir storage report (nested dictionaries)

```python
reservoirs = {
    "Reservoir A": {"capacity_af": 90000, "storage_af": 31500},
    "Reservoir B": {"capacity_af": 22500, "storage_af": 17800},
    "Reservoir C": {"capacity_af": 4800, "storage_af": 1150},
}
```

**1.** Print each reservoir's percent full using `:.1%`, and Reservoir A with thousands separators: `Reservoir A: 31,500 of 90,000 af (35.0% full)`.

```python
# Your code here
for key in reservoirs.keys():
    print(round(((reservoirs[key].get("storage_af") / reservoirs[key].get("capacity_af")) *100),1))
```

**2.** Compute the percent full of the whole system (total storage / total capacity).

```python
# Your code here
total_storage = 0
total_capacity = 0
for key in reservoirs.keys():
    total_storage += reservoirs[key].get("storage_af")
    total_capacity += reservoirs[key].get("capacity_af")
print(total_storage/total_capacity * 100)
```

```python
# Your code here
reservoirs["Reservoir C"]["storage_af"] += 250
print(reservoirs.get("Reservoir C"))
```
